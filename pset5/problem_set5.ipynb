### Question 4

For `./data/ames_prices.csv`, define an indicator equal to 1 when `Central.Air == 'Y'`. Estimate the proportion of homes with central air and test $H_0:p=.90$. Treat a proportion as the mean of a binary variable and complete the full inference routine.

```python
prices = pd.read_csv('ames_prices.csv')

# Setting 'Y' = 1 and 'N' = 0
prices['Central.Air'] = prices['Central.Air'].replace("Y", 1)
prices['Central.Air'] = prices['Central.Air'].replace("N", 0)
```

```python
estimate = prices['Central.Air'].mean() # estimate is mean of our sample in this case
print(f"estimated proportion = {estimate}")
```

We estimate that the proportion of homes with central air is about $0.933$.

#### Bootstrapping

```python
## Bootstrapping:
R = 2_000

np.random.seed(5030)    # Setting the random seed

sample_props = []

for t in range(R):
    # Generating the sample
    idx = np.random.randint(0, len(prices), len(prices))
    sample = prices.iloc[idx]

    # Proportion of 'Yes'
    o_prop = sample['Central.Air'].mean()

    sample_props.append(o_prop)
```

#### Estimating $\hat{SE}(\hat{\theta})$

```python
standard_error = np.std(sample_props, ddof=1)
print(f'standard error = {standard_error:.5f}')
```

#### Constructing 90%, 95%, 99% CIs

```python
# For 90%
lower_90 = estimate - 1.645 * standard_error
upper_90 = estimate + 1.645 * standard_error

# For 95%
lower_95 = estimate - 1.960 * standard_error
upper_95 = estimate + 1.960 * standard_error

# For 99%
lower_99 = estimate - 2.576 * standard_error
upper_99 = estimate + 2.576 * standard_error

print(f"90% CI: [{lower_90:.4f}, {upper_90:.4f}]")
print(f"95% CI: [{lower_95:.4f}, {upper_95:.4f}]")
print(f"99% CI: [{lower_99:.4f}, {upper_99:.4f}]")
```

#### Testing Hypotheses

```python
z = (estimate - 0.9)/standard_error
p = 2 * norm.cdf(-np.abs(z))

print(f"z-statistic: {z:.4f}, p-value: {p}")
```

With the calculated z-statistic, we get a p-value well below all three significance levels of $\alpha = 0.10, 0.05, 0.01$. This means at all three significance levels we reject the null hypothesis that $\mu = 0.9$.

This decision is confirmed by our confidence intervals. $0.9$ does not fall in any of the three confidence intervals, meaning these intervals also tell us that we reject the null hypothesis at $\alpha = 0.10, 0.05, 0.01$.

#### Interpretation

Our estimate approximates that the proportion of homes that have central air is about $0.933$ with about $0.00449$ standard error. Our confidence intervals tell us that we cannot reject mean values from $[0.9257, 0.9405]$ at the $\alpha = 0.10$ level, values from $[0.9243, 0.9419]$ at the $\alpha = 0.05$ level, and values from $[0.9215, 0.9447]$ at the $\alpha = 0.01$ level. If we calculated these intervals many times, approximately 90%/95%/99% of them would contain fixed, unknown proportion of homes with central air. Our hypothesis test tells us that we reject the null hypothesis and conclude that the proportion of homes with central air is not equal to $0.9$.

#### Testing Hypotheses

```python
z = estimate/standard_error
p = 2 * norm.cdf(-np.abs(z))

print(f"z-statistic: {z:.4f}, p-value: {p}")
```

With the calculated z-statistic, we get a p-value well below all three significance levels of $\alpha = 0.10, 0.05, 0.01$. This means at all three significance levels we reject the null hypothesis that $\rho = 0$.

This decision is confirmed by our confidence intervals. $0$ does not fall in any of the three confidence intervals, meaning these intervals also tell us that we reject the null hypothesis at $\alpha = 0.10, 0.05, 0.01$.

### Question 6

For `./data/metabric.csv`, use `Overall Survival (Months)` to estimate the mean recorded time to death or last follow-up, including both living and deceased patients. Let $\mu$ denote the population mean of this recorded-time variable, and test $H_0:\mu=120$ months. Complete the full inference routine and interpret the CIs in both months and years. Living patients' eventual death times are right-censored, so this exercise does not estimate unrestricted mean lifetime. Explain that distinction in your interpretation.

```python
metabric = pd.read_csv('metabric.csv')

estimate = metabric['Overall Survival (Months)'].mean()
print(f"Mean time to death (months) = {estimate}")
print(f"Mean time to death (years) = {estimate / 12}")
```

We estimate the mean time to death, including both living and deceased patients, is about 129 months or about 11 years.

#### Bootstrapping

```python
## Bootstrapping:
R = 2_000

np.random.seed(6021)    # Setting the random seed

sample_months = []
sample_years = []

for t in range(R):
    # Generating the sample
    idx = np.random.randint(0, len(metabric), len(metabric))
    sample = metabric.iloc[idx]

    # Correlation between the two features
    o_months = sample['Overall Survival (Months)'].mean()

    sample_months.append(o_months)
    sample_years.append(o_months / 12)
```

#### Estimating $SE(\hat{\theta})$

```python
standard_error_months = np.std(sample_months, ddof=1)
standard_error_years = np.std(sample_years, ddof=1)
print(f'standard error months = {standard_error_months:.5f}')
print(f'standard error years = {standard_error_years:.5f}')
```

#### Constructing 90%, 95%, 99% CIs

```python
estimate_years = estimate / 12

# For 90%
lower_90_months = estimate - 1.645 * standard_error_months
upper_90_months = estimate + 1.645 * standard_error_months
lower_90_years = estimate_years - 1.645 * standard_error_years
upper_90_years = estimate_years + 1.645 * standard_error_years

# For 95%
lower_95_months = estimate - 1.960 * standard_error_months
upper_95_months = estimate + 1.960 * standard_error_months
lower_95_years = estimate_years - 1.960 * standard_error_years
upper_95_years = estimate_years + 1.960 * standard_error_years

# For 99%
lower_99_months = estimate - 2.576 * standard_error_months
upper_99_months = estimate + 2.576 * standard_error_months
lower_99_years = estimate_years - 2.576 * standard_error_years
upper_99_years = estimate_years + 2.576 * standard_error_years

print(f"90% CI months: [{lower_90_months:.4f}, {upper_90_months:.4f}]")
print(f"90% CI years: [{lower_90_years:.4f}, {upper_90_years:.4f}]")
print()
print(f"95% CI months: [{lower_95_months:.4f}, {upper_95_months:.4f}]")
print(f"95% CI years: [{lower_95_years:.4f}, {upper_95_years:.4f}]")
print()
print(f"99% CI months: [{lower_99_months:.4f}, {upper_99_months:.4f}]")
print(f"99% CI years: [{lower_99_years:.4f}, {upper_99_years:.4f}]")
```

#### Testing Hypotheses

```python
z = (estimate - 120)/standard_error_months
p = 2 * norm.cdf(-np.abs(z))

z_years = (estimate_years - 10)/standard_error_years
p_years = 2 * norm.cdf(-np.abs(z_years))

print(f"z-statistic: {z:.4f}, p-value: {p}")
print(f"z-statistic: {z_years:.4f}, p-value: {p_years}")
```

With the calculated z-statistic, we get a p-value well below all three significance levels of $\alpha = 0.10, 0.05, 0.01$ for both months and years. This means at all three significance levels we reject the null hypothesis that $\mu = 120$ for months and $10$ for years.

This decision is confirmed by our confidence intervals. $120$ months and $10$ years both do not fall in any of their three corresponding confidence intervals, meaning these intervals also tell us that we reject the null hypothesis at $\alpha = 0.10, 0.05, 0.01$.

#### Interpretation

Our estimate approximates that the mean time to death, including both living and deceased patients, is about 129 months or about 11 years with about $2.16791$ and $0.18066$ standard error respectively. Our confidence intervals tell us that we cannot reject mean values from $[125.5281, 132.6605]$ for months and $[10.4607, 11.0550]$ for years at the $\alpha = 0.10$ level, values from $[124.8452, 133.3434]$ for months and $[10.4038, 11.1120]$ for years at the $\alpha = 0.05$ level, and values from $[123.5097, 134.6788]$ for months and $[10.2925, 11.2232]$ for years at the $\alpha = 0.01$ level. If we calculated these intervals many times, approximately 90%/95%/99% of them would contain fixed, unknown mean time to death (months or years). Our test hypothesis test tells us that we reject the null hypothesis and conclude that the mean time to death is not equal to 120 months or 10 years.

Since this data is right-censored, we have not observed the death of some of the patients. We cannot use this test to estimate unrestricted mean lifetime because not all of our data is the definitive time of death for some patients, thus cannot be representative of the population.

Do not interpret these results as nationally representative population estimates or survey-adjusted confidence intervals.

```python
nhanes = pd.read_csv('nhanes_data_17_18.csv')
older18 = nhanes.loc[nhanes['AgeInYearsAtScreening'] >= 18]
sleep = older18['SleepHoursWeekdaysOrWorkdays']
sleep = sleep.dropna()

estimate = sleep.mean()
print(f"Mean weekday sleep hours = {estimate:.4f}")
```

We estimate that the average weekday sleep hours for respondents 18 or older is 7.6212.

#### Bootstrapping

```python
## Bootstrapping:
R = 2_000

np.random.seed(5012)    # Setting the random seed

sample_means = []

for t in range(R):
    # Generating the sample
    idx = np.random.randint(0, len(sleep), len(sleep))
    sample = sleep.iloc[idx]

    # Proportion of 'Yes'
    o_mean = sample.mean()

    sample_means.append(o_mean)
```

#### Estimating $SE(\hat{\theta})$

```python
standard_error = np.std(sample_means, ddof=1)
print(f'standard error = {standard_error:.5f}')
```

#### Constructing 90%, 95%, 99% CIs

```python
# For 90%
lower_90 = estimate - 1.645 * standard_error
upper_90 = estimate + 1.645 * standard_error

# For 95%
lower_95 = estimate - 1.960 * standard_error
upper_95 = estimate + 1.960 * standard_error

# For 99%
lower_99 = estimate - 2.576 * standard_error
upper_99 = estimate + 2.576 * standard_error

print(f"90% CI: [{lower_90:.4f}, {upper_90:.4f}]")
print(f"95% CI: [{lower_95:.4f}, {upper_95:.4f}]")
print(f"99% CI: [{lower_99:.4f}, {upper_99:.4f}]")
```

#### Test Hypotheses

```python
z = (estimate - 8)/standard_error
p = 2 * norm.cdf(-np.abs(z))

print(f"z-statistic: {z:.4f}, p-value: {p}")
```

With the calculated z-statistic, we get a p-value below all three significance levels of $\alpha = 0.10, 0.05, 0.01$. This means at all three significance levels we reject the null hypothesis that $\mu = 8$ hours.

This decision is confirmed by our confidence intervals. $8$ does not fall in any of the three confidence intervals, meaning these intervals also tell us that we reject the null hypothesis at $\alpha = 0.10, 0.05, 0.01$.

#### Interpretation

Our estimate approximates that the average weekday sleep hours for respondents 18 or older is about $7.6212$ with about $0.02255$ standard error. Our confidence intervals tell us that we cannot reject mean values from $[7.5842, 7.6583]$ at the $\alpha = 0.10$ level, values from $[7.5770, 7.6654]$ at the $\alpha = 0.05$ level, and values from $[7.5632, 7.6793]$ at the $\alpha = 0.01$ level. Our test hypothesis test tells us that we reject the null hypothesis and conclude that the average weekday sleep hours is not equal to $8$.
