```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

rng = np.random.default_rng(20260717)
df = pd.read_csv('metabric.csv')
```

```python
def ecdf_xy(samples):
    x = np.sort(np.asarray(samples))
    y = np.arange(1, len(x) + 1) / len(x)
    return x, y
```

```python
df
```

```python
x, y = ecdf_xy(df['Overall Survival (Months)'])
plt.step(x, y, where='post', color='blue', lw=2)
```

```python
chemo = df.loc[df['Chemotherapy'] == 'YES']
no_chemo = df.loc[df['Chemotherapy'] == 'NO']

fig, ax = plt.subplots(figsize=(5, 4))
x, y = ecdf_xy(chemo['Overall Survival (Months)'])
ax.step(x, y, where='post', color='blue', lw=2, label="Chemo")

x, y = ecdf_xy(no_chemo['Overall Survival (Months)'])
ax.step(x, y, where='post', color='orange', lw=2, label="No Chemo")
ax.legend()
plt.show()
```
